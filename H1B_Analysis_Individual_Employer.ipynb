### <font size = 2> For 'Raw' dropdown selection, the wage_range_d_flag is disabled. 

If 'Filtered' is selected, users can check or uncheck 'wage_range_d_flag' based on their analysis range of values.  
* If 'wage_range_d_flag' is selected, the range for analysis is taken from the fixed dictionary namely 'wage_range_dict_p' & 'wage_range_dict_l'. 
* If 'wage_range_d_flag' is not selected, the range for analysis is set in the code in a tuple format. Refer to 'else' part of the code.  

If 'ALL' is selected, it lists out 'Top N Employers' for that year. 'Top N Employers' value can be changed adjusting the slider. 'N' is the value at the slider poistion
    To generate PDF set generate_pdf=True in H1BAnalysis.analysis()
    To display the graphs in the Jupyter, set generate_pdf=False, which by default is False.

```python
lwr_list = list(map(lambda x: str(x)+'%', range(10, 51, 5)))
# lwr_list.insert(0, '5%')
pwr_list = list(map(lambda x: str(x)+'%', range(125, 525, 25)))
# pwr_list.insert(0, '5%')

employer_name = widgets.Dropdown(options = dropdown_list, description = 'Employer Name: ', layout={'width': '300px'}, style={'description_width': 'auto'})
data_type = widgets.Dropdown(options = ['Raw', 'Filtered'], value = 'Raw', description = 'Analysis: ', layout={'width': '200px'})
# data_type = widgets.Dropdown(options = ['Raw', 'Filtered'], layout={'width': '100px'})
year = widgets.Dropdown(options = range(2008, 2020), layout={'width': '200px'}, description = 'Year: ')
# topne = widgets.IntSlider(value = 10, min = 5, max = 100, step = 5, description = 'Top N Employers:', style={'description_width': 'auto'}, layout={'width': '400px'})
wage_range_d_flag = widgets.Checkbox(value  = False, description = 'wage_range_dict_flag' , disabled= True, continuous_update=False)
lower_limit = widgets.Dropdown(options = lwr_list, description = 'Lower_Limit_Wage_Range', style={'description_width': 'auto'}, layout={'width': 'auto'}, disabled = True)
higher_limit = widgets.Dropdown(options = pwr_list, description = 'Higher_Limit_Wage_Range', style={'description_width': 'auto'}, layout={'width': 'auto'}, disabled = True)
wage_range_by_median = widgets.Checkbox(value  = False, description = 'wage_range_by_median' , disabled= True, continuous_update=False, layout={'width': 'auto'},)
output_pdf = widgets.Checkbox(value  = False, description = 'generate_pdf' , continuous_update=False, layout={'width': 'auto'},)

btn = widgets.Button(description = 'Run', layout={'width': '100px'})
out = widgets.Output()
def handler(c):
    with out:
        out.clear_output()
        if (year.value == 2009):
            return(display('2009 H1B data not captured/present in the selected database'))
        else:
            df_y = df[(df.YEAR == year.value)]
            
            if (employer_name.value != 'ALL'):
                employer_name_str = str('^') + str(employer_name.value)
                # this is for getting range limits
                df_p = df_y[df_y.EMPLOYER_NAME.str.contains(employer_name_str, case = False) == True]
                if (data_type.value == 'Raw'):
                        analysis(df_y, bins = 30, data_type = data_type.value, employer_name = employer_name.value, 
                                                   year = year.value, generate_pdf=output_pdf.value)
                elif (data_type.value == 'Filtered'):
                    if (wage_range_by_median.value == True):
                        ll_f = eval(lower_limit.value.rstrip('%'))/100
                        hl_m = eval(higher_limit.value.rstrip('%'))/100
                        range_pw = (df_p.PREVAILING_WAGE.median()*ll_f, df_p.PREVAILING_WAGE.median()* hl_m)
                        range_lw = (df_p.LCA_WAGE.median() * ll_f, df_p.LCA_WAGE.median()* hl_m )
                        analysis(df_y, bins = 30, data_type = data_type.value, employer_name = employer_name.value, 
                                                   year = year.value, pw_range = range_pw, lw_range = range_lw, generate_pdf=output_pdf.value)
                        # display(range_pw, range_lw)
                    elif (wage_range_d_flag.value == True):
                            analysis(df_y, bins = 30, data_type = data_type.value, employer_name = employer_name.value, 
                                                       year = year.value, pw_range = wage_range_dict_p, lw_range = wage_range_dict_l,
                                                       dataframe_checked=True, generate_pdf=output_pdf.value)
                    else:
                        analysis(df_y, bins = 30, data_type = data_type.value, year = year.value, employer_name = employer_name.value, 
                                                   pw_range = (20000, 150000), lw_range = (30000, 250000), generate_pdf=output_pdf.value)
                else:
                    display('data_type.value must be either \'Raw\' or \'Filtered\'')
            else:
                display('Select an employer')
        
def wage_range_select_status(x):
    if (x == 'Raw'):
        wage_range_d_flag.value = False
        wage_range_by_median.value = False
        wage_range_by_median.disabled = True
        wage_range_d_flag.disabled = True
        lower_limit.disabled = True
        higher_limit.disabled = True
    elif(x == 'Filtered'):
        wage_range_d_flag.disabled = False
        wage_range_by_median.disabled = False
        wage_range_by_median.value = True
        wage_range_d_flag.value = False
    else:
        # do nothing..
        print('Check dropdown flag.. ')
        
def wage_range_select_by_median(x):
    if(x == False):
        print(x)
        lower_limit.value = 'default'
        higher_limit.value = 'default'
        wage_range_d_flag.value = False
        lower_limit.disabled = True
        higher_limit.disabled = True
        wage_range_d_flag.disabled = False
        wage_range_d_flag.value = True
    else:
        lower_limit.value = lwr_list[4]
        higher_limit.value = pwr_list[7]
        wage_range_d_flag.value = False
        lower_limit.disabled = False
        higher_limit.disabled = False
        wage_range_d_flag.disabled = False
        wage_range_d_flag.value = False
        
def wage_range_select_dict_flag(x):
    if (x == False):
        wage_range_by_median.value = True
        wage_range_by_median.disabled = False
    else:
        wage_range_by_median.value = False
        wage_range_by_median.disabled = False
        lower_limit.disabled = True
        higher_limit.disabled = True
        
btn.on_click(handler)
interactive(wage_range_select_status, x = data_type);
interactive(wage_range_select_by_median, x = wage_range_by_median);
interactive(wage_range_select_dict_flag, x = wage_range_d_flag);
# interactive(output_pdf, x = output_pdf);
widgets.VBox([widgets.HBox(children = (data_type, output_pdf)), widgets.HBox(children = (wage_range_by_median, lower_limit, higher_limit, wage_range_d_flag)), widgets.HBox([year, employer_name, btn]), out])
```
