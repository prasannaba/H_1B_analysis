```python
if os.path.exists("DataFrame_top30_Employers_Median_Final.csv"):
    print("File is present at the path, reading it...")
    df_median = pd.read_csv(
        "DataFrame_top30_Employers_Median_Final.csv", low_memory=False, memory_map=True
    )
else:
    print("File is not present at the path, so create it...")
    df_median = pd.read_csv(
        "DataFrame_top30_Employers_Final.csv", low_memory=False, memory_map=True
    )
    for emp in df_median["EMPLOYER_NAME"].unique().tolist():
        for year in df_median["YEAR"].unique().tolist():
            # print(emp, year)
            mask = (df_median["EMPLOYER_NAME"] == emp) & (df_median["YEAR"] == year)
            pw_mask = mask & (df_median["PREVAILING_WAGE"] < 35000)  # noqa: PLR2004
            lc_mask = mask & (df_median["LCA_WAGE"] < 35000)  # noqa: PLR2004
            pw_median, lc_median = (
                df_median[mask][["PREVAILING_WAGE", "LCA_WAGE"]].median().tolist()
            )
            if pw_median < 35000:  # noqa: PLR2004
                pw_median = lc_median
            df_median.loc[pw_mask, "PREVAILING_WAGE"] = pw_median
            df_median.loc[lc_mask, "LCA_WAGE"] = lc_median
            df_median.to_csv("DataFrame_top30_Employers_Median_Final.csv", index=False)
```

```python
lwr_list = list(map(lambda x: str(x) + "%", range(10, 51, 5)))
lwr_list.insert(0, "default")
pwr_list = list(map(lambda x: str(x) + "%", range(125, 525, 25)))
pwr_list.insert(0, "default")

employer_name = widgets.Dropdown(
    options=dropdown_list,
    description="Employer Name: ",
    layout={"width": "300px"},
    style={"description_width": "auto"},
)
wage_range_d_flag = widgets.Checkbox(value=False, description="wage_range_dict_flag")

year_range = widgets.IntRangeSlider(
    value=[2008, 2019],
    min=2008,
    max=2019,
    step=1,
    continuous_update=False,
    description="Range of Year: ",
    layout={"width": "400px"},
    style={"description_width": "auto"},
)

data_type = widgets.Dropdown(
    options=["Raw", "Filtered"], description="Analysis:", value="Filtered"
)
lower_limit = widgets.Dropdown(
    options=lwr_list,
    value="default",
    description="Lower_Limit_Wage_Range",
    style={"description_width": "auto"},
    layout={"width": "auto"},
    disabled=False,
)
higher_limit = widgets.Dropdown(
    options=pwr_list,
    value="default",
    description="Higher_Limit_Wage_Range",
    style={"description_width": "auto"},
    layout={"width": "auto"},
    disabled=False,
)
wage_range_by_median = widgets.Checkbox(
    value=False,
    description="wage_range_by_median",
    disabled=False,
    continuous_update=False,
)

btn = widgets.Button(description="Run")
out = widgets.Output()


def handler(c):
    with out:
        out.clear_output()
        # clear_output()
        if year_range.value[0] > year_range.value[1]:
            return "Lower Range Limit 'm': {} should be less than Higher Range Limit 'n': {}".format(
                year_range.value[0], year_range.value[1]
            )

        wage_range_dict_p = {}
        wage_range_dict_l = {}
        if employer_name.value != "ALL":
            if data_type.value == "Filtered":
                if not wage_range_d_flag.value:
                    # select generic
                    wage_range_dict_p = wage_range_dict_p_g
                    wage_range_dict_l = wage_range_dict_l_g
                elif wage_range_d_flag.value == True:
                    # if one of the limit is 'default', select generic
                    if (lower_limit.value == "default") or (
                        higher_limit.value == "default"
                    ):
                        # select generic
                        wage_range_dict_p = wage_range_dict_p_g
                        wage_range_dict_l = wage_range_dict_l_g
                    else:
                        # select employer & calculate the ranges for years selected
                        df_dict = df[df.EMPLOYER_NAME == employer_name.value]
                        wage_range_dict_t_p = {}
                        wage_range_dict_t_l = {}
                        for i in range(year_range.value[0], year_range.value[1] + 1):
                            if i == 2009:
                                continue
                            df_y = df_dict[df_dict.YEAR == i]
                            ll_f = eval(lower_limit.value.rstrip("%")) / 100
                            hl_m = eval(higher_limit.value.rstrip("%")) / 100
                            range_pw = (
                                df_y.PREVAILING_WAGE.median()
                                - (df_y.PREVAILING_WAGE.median() * ll_f),
                                df_y.PREVAILING_WAGE.median()
                                + (df_y.PREVAILING_WAGE.median() * hl_m),
                            )
                            range_lw = (
                                df_y.LCA_WAGE.median()
                                - (df_y.LCA_WAGE.median() * ll_f),
                                df_y.LCA_WAGE.median()
                                + (df_y.LCA_WAGE.median() * hl_m),
                            )
                            wage_range_dict_t_p.update({i: range_pw})
                            wage_range_dict_t_l.update({i: range_lw})

                        wage_range_dict_p = wage_range_dict_t_p
                        wage_range_dict_l = wage_range_dict_t_l
                else:
                    display("Check the options")
            elif data_type.value == "Raw":
                display(
                    "For raw data analysis, range limit is not applicable & is empty, the 'wage_range_dict_flag' is cleared & disabled"
                )
                # wage_range_dict_p & wage_range_dict_l are null & because of the wage_range_dict_flag is cleared are not used in the library.
            else:
                display("Select raw or filtered")

            df_range1 = pd.DataFrame.from_dict(
                wage_range_dict_p,
                orient="index",
                columns=["PW_LOWER_LIMIT", "PW_UPPER_LIMIT"],
            )
            df_range2 = pd.DataFrame.from_dict(
                wage_range_dict_l,
                orient="index",
                columns=["LCA_LOWER_LIMIT", "LCA_UPPER_LIMIT"],
            )
            df_range = (
                pd.concat([df_range1, df_range2], sort=False, axis=1)
                .reset_index()
                .rename(columns={"index": "YEAR"})
            )
            display("The following wage range selected for analysis:")
            display(df_range)

            analysis2(
                df,
                employer_name.value,
                year_range.value[0],
                year_range.value[1],
                wage_range_dict_flag=wage_range_d_flag.value,
                pw_range=wage_range_dict_p,
                lw_range=wage_range_dict_l,
                dataframe_checked=True,
                generate_pdf=True,
            )
        else:
            display("Select a specific employer name from drop down list")


def wage_range_select_status(x):
    if x == "Raw":
        wage_range_d_flag.value = False
        wage_range_by_median.value = False
        wage_range_d_flag.disabled = True
        wage_range_by_median.disabled = True
        lower_limit.disabled = True
        higher_limit.disabled = True
    else:
        wage_range_d_flag.disabled = False
        wage_range_d_flag.value = True
        wage_range_by_median.disabled = False


def wage_range_select_dict_flag(x):
    if not x:
        lower_limit.value = "default"
        higher_limit.value = "default"
        wage_range_d_flag.value = False
        lower_limit.disabled = True
        higher_limit.disabled = True
        wage_range_d_flag.disabled = False
    else:
        # lower_limit.value = lwr_list[5]
        # higher_limit.value = pwr_list[8]
        wage_range_d_flag.value = True
        lower_limit.disabled = False
        higher_limit.disabled = False
        wage_range_d_flag.disabled = False


btn.on_click(handler)
interact(wage_range_select_status, x=data_type)
interactive(wage_range_select_dict_flag, x=wage_range_d_flag)

widgets.VBox(
    [
        widgets.HBox(children=(wage_range_d_flag, lower_limit, higher_limit)),
        widgets.HBox([employer_name, year_range, btn]),
        out,
    ]
)
```
